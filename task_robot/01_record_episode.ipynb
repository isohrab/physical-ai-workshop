# LeRobot Workshop — From Demonstration to Robot Autonomy

Today you'll walk the whole path a real robot-learning project takes:

**Record demonstrations → build a dataset → train a policy → run it on the robot**

We won't have GPUs at every seat, so the plan is:

1. See how a dataset gets recorded (with LeLab) — conceptually, so you understand where the data comes from.
2. Explore a dataset we already recorded — a real one, public on the Hugging Face Hub.
3. See, in broad strokes, how that dataset became a trained policy (SmolVLA) — again, we already did the actual training.
4. Connect to a shared GPU and run that trained policy on your own robot arm, live.

Then: a challenge, in a separate notebook, where you build something on top of it yourselves.

## Part 1 — Recording your own dataset (with LeLab)

This is the part of the pipeline that happens *before* any of today's shared resources exist. If you had your own robot arm and camera at home, here's the shape of it, broken into 5 steps — **what** to do and the exact commands, not just the idea (LeLab's own guided flow handles the fine detail of each step):

### Step 1 — Install LeLab

Run this in a **terminal**, inside a **fresh virtual environment** dedicated to LeLab — it pulls in a specific pinned version of `lerobot` from GitHub, so keep it separate from any other `lerobot` environment you set up for Part 4 (`03_training_inference.ipynb`):

```bash
python3 -m venv lelab-env
source lelab-env/bin/activate      # Windows: lelab-env\Scripts\activate

git clone https://github.com/huggingface/lelab.git
cd lelab
pip install -e .
```

Requires Python ≥3.10. `pip install -e .` also pulls in the matching `lerobot` version — you don't install that separately.

### Step 2 — Launch LeLab and connect both arms

Plug in the leader arm, the follower arm (the one that moves), and the camera via USB *first*, then, in the same terminal/venv as Step 1:

```bash
lelab
```

This starts LeLab and opens its UI in your browser automatically.

### Step 3 — Calibrate each arm

In the browser UI, open **Calibrate** and follow the guided flow — once for the leader, once for the follower. LeLab tells you exactly what to move and when.

⚠️ **Write down** the `robot_port`, calibration `id`, and `camera_index` LeLab shows you here for the follower arm — you'll need these exact values again in Part 4 (`03_training_inference.ipynb`).

### Step 4 — Start a recording session 

In the **Record** tab, give it:
- a task description in plain English (e.g. *"pick up the letter M and place it in the first box"*)
- how many episodes to record
- how long each episode should run


### Step 5 — Perform the task, repeatedly, and let LeLab save it

Do the task with the leader arm, the same way, several times. Consistency matters far more than speed — the policy will learn *your* pattern. When you're done, LeLab writes the episodes into a dataset on disk, and can push it straight to the Hugging Face Hub with one click in the **Upload** tab.

That's genuinely it — no code required for this part. We're not recording live today so everyone gets to the same starting point; instead, in Part 2 we'll open a dataset recorded exactly this way.
