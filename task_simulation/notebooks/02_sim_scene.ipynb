## What's in the scene

One USD file (`scene.usd`) holding a small kitchen:

- **The room** — floor, walls, and their backing geometry.
- **Fixtures** — counters, a fridge housing, cabinets, and stacked units.
- **The task props** — three oranges (`Orange001`, `Orange002`, `Orange003`) and a `Plate`,
  sitting on the counter in front of where the robot is placed.

The SO-101 arm is *not* part of the scene USD. It's a separate asset
(`assets/robots/so101_follower.usd`) that the task config drops into the world — which is why the
same kitchen can be reused for different robots and different tasks.

## From static geometry to physics objects

By default the whole USD loads as one **static** asset — nice
to look at, but the oranges wouldn't be grabbable and nothing could query where they are.

The task config fixes that with one line:

```python
parse_usd_and_create_subassets(
    KITCHEN_WITH_ORANGE_USD_PATH,
    self,
    specific_name_list=['Orange001', 'Orange002', 'Orange003', 'Plate'],
)
```

That promotes those four named prims into individual **rigid bodies** in the scene. Once they are
rigid bodies, two things become possible:

1. The gripper can actually collide with and lift them.
2. The task code can ask for `Orange001.data.root_pos_w` — the position readout that the success
   conditions in the next notebook are built on.

Everything *not* in that list (walls, counters, the fridge) stays static scenery. That's
deliberate: fewer simulated bodies means a faster, more stable sim.
