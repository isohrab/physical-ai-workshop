## How the scene was created

For context, the scene was built with the following steps:

1. Opened `kitchen_with_orange/scene.usd` in Isaac Sim and removed the orange objects from the **Stage** tree.

2. Opened **Window → Browsers → NVIDIA Assets**, searched for `MechanicalPencil`, and dragged the pencil onto the counter near the plate.

   ![NVIDIA Assets browser showing the MechanicalPencil search](resources/screenshot2.png)

3. Applied the **Rigid Body with Colliders Preset** to the pencil. This allows the robot to pick it up and ensures that it physically interacts with objects such as the plate.

4. Used **File → Collect and Save As** to save the complete scene and its required assets as `kitchen_with_pen`.

   ![Kitchen scene with the pen positioned on the counter](resources/screenshot3.png)
