## What scene are we using?

The task is built on a kitchen scene originally created for an orange-picking demo (`kitchen_with_orange`) from [LeIsaac](https://github.com/LightwheelAI/leisaac). For this workshop the oranges were swapped for a pen, so the robot's job becomes: **pick up the pen and place it on the plate**.

![Kitchen with oranges scene](resources/screenshot1.png)

## How `kitchen_with_pen` was built

You don't need to redo any of this — the finished scene already lives in `~/workspace/leisaac/assets/scenes/kitchen_with_pen/`. For context, here's how it was created:

1. Opened `kitchen_with_orange/scene.usd` in Isaac Sim and deleted the orange objects from the Stage tree.
2. Opened **Window → Browsers → NVIDIA Assets**, searched for "MechanicalPencil", and dragged it onto the counter near the plate.

   ![NVIDIA Assets browser with MechanicalPencil search](resources/screenshot2.png)

3. Added a **Rigid Body with Colliders Preset** to the pencil so the robot can pick it up and it physically interacts with the plate.
4. Used **File → Collect and Save As** to bundle the scene and its assets into `kitchen_with_pen`.

   ![Kitchen scene with pen positioned on counter](resources/screenshot3.png)
