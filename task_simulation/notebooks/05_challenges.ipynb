# Workshop Challenges

## Introduction

You just watched a fine-tuned GR00T N1.5 policy drive the arm. So — how did it do?

- Did it get all three oranges onto the plate, or did it stall after the first?
- When it failed, *what* failed: the reach, the grasp, or the release?
- The oranges, plate, and front camera are re-randomised every round (±3 cm, ±2.5°). Did the
  rounds it failed on look different from the ones it aced?

Share your observations with us :)

## Challenge 1: Manual keyboard control

Drive the arm yourself and get the oranges onto the plate. It's harder than it looks — and it's
a fast way to appreciate what the policy is doing.

```bash
conda activate leisaac
cd ~/workspace/leisaac

python scripts/environments/teleoperation/teleop_se3_agent.py \
    --task=LeIsaac-SO101-PickOrange-v0 \
    --teleop_device=keyboard \
    --num_envs=1 \
    --device=cuda \
    --enable_cameras
```

Once the Isaac Lab window has focus:

- press **`b`** to begin teleoperation,
- press **`n`** to reset and mark the attempt a **success**,
- press **`r`** to reset and mark it a **failure**.

Add `--record --dataset_file=./datasets/my_demos.hdf5` and you're doing exactly what produced the
source demonstrations for this task's training data. Get ten good episodes and you have the input
Isaac Lab Mimic needs (see the mimic section of [notebook 3](03_sim_task.ipynb)).

## Challenge 2: RL training policy

Everything you saw today was **imitation** learning — a policy copying human demonstrations. Define
a reward function and an RL training setup in Isaac Lab for this task instead, and train a policy
from scratch in the simulator. Note that `PickOrangeEnvCfg` ships with an empty `RewardsCfg` — that
class is where you'd start.

## Challenge 3: Gripper enhancement

If you saw an orange slip or roll out of the jaws, that's a *physics* problem, not a policy
problem. Add rubber tips to the SO-101 gripper — a physics material with higher friction on the jaw
surfaces — and see whether the same policy suddenly does better. (You can find the answer ;)
